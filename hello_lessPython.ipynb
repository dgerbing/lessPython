# hello, lessPython

A minimal check that the published package works: read a bundled dataset, then make a `Chart()`.

Run each cell with **Shift+Enter**. Make sure the kernel (top-right) is **Python (lesspy)**.

```python
import lessPy as lp
import plotly.io as pio
pio.renderers.default = "iframe".
```

```python
# read a bundled example dataset (returns a pandas DataFrame)
d = lp.read_data("Employee")
d.head()
```

```python
# bar chart of counts for a categorical variable
# variables are passed as strings naming columns of d
lp.Chart("Dept", data=d)
```
